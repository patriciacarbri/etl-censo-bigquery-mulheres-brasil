# Análise Exploratória de Dado (EDA)

```python
# Bibliotecas necessárias
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Importação dos dados
caminho_dados = '../data/raw/bq_2022.csv'
df = pd.read_csv(caminho_dados, delimiter=',', low_memory=False)
```

```python
print("\n--- Primeiras linhas do Dataframe ---")
df.head()
```

```python
print("\n--- Informações das Colunas ---")
df.info()
```

```python
if 'df' in locals():
    # 4. Verificar a quantidade de dados faltantes por coluna
    print("\n--- Contagem de Dados Faltantes (NaN) por Coluna ---")
    dados_faltantes = df.isnull().sum()
    
    if dados_faltantes.sum() == 0:
        print("Não há dados faltantes no DataFrame.")
    else:
        print(dados_faltantes[dados_faltantes > 0])
```

```python
print("\n--- Descrição dos Dados Numéricos ---")
df.describe()
```
